```python
def check_grammar(text):
    errors = grammar_tool.check(text)
    corrected_text = language_tool_python.utils.correct(text, errors)
    return corrected_text

print("Grammar and spelling checker initialized successfully!")
```

```python
sample_text = "Artificial intelligence are changing the world. It help people to learn faster."

corrected_text = check_grammar(sample_text)

print("Original Text:", sample_text)
print("Corrected Text:", corrected_text)
```

```python
def ai_paraphrasing_tool(text):
    paraphrased = paraphrase_text(text)
    corrected = check_grammar(paraphrased)
    return corrected

print("AI-Powered Paraphrasing Tool is ready and returning corrected text!")
```

```python
import nltk
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from rouge_score import rouge_scorer
from sentence_transformers import SentenceTransformer, util


nltk.download('punkt', quiet=True)


similarity_model = SentenceTransformer('all-MiniLM-L6-v2')


original_text = "Artificial intelligence is changing the way people learn and work."

reference_text = "AI is transforming how individuals study and perform their jobs."


generated_text = ai_paraphrasing_tool(original_text)


bleu_score = sentence_bleu(
    [reference_text.split()],
    generated_text.split(),
    smoothing_function=SmoothingFunction().method1
)


scorer = rouge_scorer.RougeScorer(['rouge1', 'rougeL'], use_stemmer=True)

rouge_results = scorer.score(reference_text, generated_text)


embeddings = similarity_model.encode(
    [original_text, generated_text],
    convert_to_tensor=True
)

similarity_score = util.cos_sim(
    embeddings[0], embeddings[1]
).item()


print("EVALUATION RESULTS")
print("-" * 40)

print("Original Text:", original_text)
print("Reference Text:", reference_text)
print("Generated Text:", generated_text)

print("\nBLEU Score:", round(bleu_score, 4))
print("ROUGE-1 Score:", round(rouge_results['rouge1'].fmeasure, 4))
print("ROUGE-L Score:", round(rouge_results['rougeL'].fmeasure, 4))
print("Semantic Similarity:", round(similarity_score, 4))
```

```python



def paraphrase_text(text):

    input_text = "paraphrase: " + text + " </s>"

    inputs = tokenizer(
        input_text,
        return_tensors="pt",
        max_length=512,
        truncation=True
    )

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_length=128,
            num_beams=5,
            num_return_sequences=5,
            do_sample=True,
            top_k=120,
            top_p=0.98,
            temperature=1.5,
            repetition_penalty=1.2
        )

    candidates = [
        tokenizer.decode(output, skip_special_tokens=True).strip()
        for output in outputs
    ]

    # Select a different output if available
    original = text.strip().lower()

    for candidate in candidates:
        if candidate.lower().strip(" .!") != original.strip(" .!"):
            return candidate

    return candidates[0]


print("Multiple paraphrase generation is ready!")
```

```python
test_sentence = "Artificial intelligence is changing the way people learn and work."

print("Original:", test_sentence)
print("Paraphrased:", paraphrase_text(test_sentence))
```
